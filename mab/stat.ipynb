```python
import pandas as pd
import numpy as np
from scipy.stats import friedmanchisquare, wilcoxon, spearmanr
import scikit_posthocs as sp

# =====================================
# UČITAVANJE PODATAKA
# =====================================

file_path = ".././fuzzy_outputs/Final fuzzy in the 1st/Total Score.xlsx"  # <-- promeni ime fajla

df = pd.read_excel(file_path)

# Strategije
strategies = df["Name"].values

# Uzimamo samo kolone sa Total Score
score_cols = [col for col in df.columns if "Total Score" in col]

data_matrix = df[score_cols].values  # strategije x turniri

print(f"Broj strategija: {data_matrix.shape[0]}")
print(f"Broj turnira: {data_matrix.shape[1]}\n")

# =====================================
# FRIEDMAN TEST
# =====================================

# friedmanchisquare traži svaku kolonu kao poseban argument
stat, p_value = friedmanchisquare(
    *[data_matrix[:, i] for i in range(data_matrix.shape[1])]
)

print("=== FRIEDMAN TEST ===")
print(f"Statistic = {stat:.4f}")
print(f"p-value  = {p_value:.6f}")

# =====================================
# KENDALL'S W
# =====================================

n = data_matrix.shape[1]  # broj turnira
k = data_matrix.shape[0]  # broj strategija

kendall_w = stat / (n * (k - 1))

print("\n=== KENDALL'S W ===")
print(f"W = {kendall_w:.4f}")

# =====================================
# NEMENYI POST-HOC
# =====================================

# print("\n=== NEMENYI POST-HOC (p-vrednosti) ===")

# nemenyi = sp.posthoc_nemenyi_friedman(data_matrix)

# nemenyi.index = strategies
# nemenyi.columns = strategies

# print(nemenyi)

# =====================================
# WILCOXON: FUZZY vs OSTALI
# =====================================

print("\n=== WILCOXON: Fuzzy vs ostali ===")

fuzzy_index = list(strategies).index("Fuzzy")
fuzzy_scores = data_matrix[fuzzy_index]

for i, name in enumerate(strategies):
    if name == "Fuzzy":
        continue
    
    stat_w, p_w = wilcoxon(fuzzy_scores, data_matrix[i])
    print(f"Fuzzy vs {name}: p = {p_w:.6f}")

# =====================================
# SPEARMAN korelacija (primer)
# =====================================

print("\n=== SPEARMAN korelacija ===")

# Prosečni score po strategiji
avg_scores = data_matrix.mean(axis=1)

# Korelacija sa rangom (da vidiš da li je stabilno)
ranks = avg_scores.argsort().argsort()

rho, p_corr = spearmanr(avg_scores, ranks)

print(f"Spearman rho = {rho:.4f}, p = {p_corr:.6f}")
```
