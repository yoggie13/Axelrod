```python
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import numpy as np
from matplotlib import pyplot as plt
```

```python
cooperation = ctrl.Antecedent(np.arange(0, 100, 1), 'cooperation')
adaptivity = ctrl.Antecedent(np.arange(0, 100, 1), 'adaptivity')
forgiveness = ctrl.Antecedent(np.arange(0, 100, 1), 'forgiveness')
stochastic = ctrl.Antecedent(np.arange(0, 100, 1), 'stochastic')

cooperation.automf(names=["low", "medium", "high"])
adaptivity.automf(names=["no", "yes"])
forgiveness.automf(names=["low", "medium", "high"])
forgiveness['low'] = fuzz.gaussmf(forgiveness.universe, 0, 25)
forgiveness['medium'] = fuzz.trimf(forgiveness.universe, [25, 50, 75])
stochastic.automf(names=["none", "sometimes", "always"])

resulting_strategy = ctrl.Consequent(np.arange(0, 100, 1),'resulting_strategy')
resulting_strategy['D'] = fuzz.trimf(resulting_strategy.universe, [0, 8, 9])
resulting_strategy['C'] = fuzz.trimf(resulting_strategy.universe, [59, 60, 99])
```

```python
plt.figure()
for label in cooperation.terms:
    plt.plot(cooperation.universe, cooperation[label].mf, label=label)

plt.title('Membership Functions for Input "Cooperation"')
plt.xlabel('Cooperation level')
plt.ylabel('Membership Degree')
plt.legend()
plt.show()
```

```python
plt.figure()
for label in adaptivity.terms:
    plt.plot(adaptivity.universe, adaptivity[label].mf, label=label)

plt.title('Membership Functions for Input "Adaptivity"')
plt.xlabel('Adaptivity level')
plt.ylabel('Membership Degree')
plt.legend()
plt.show()
```

```python
plt.figure()
for label in forgiveness.terms:
    plt.plot(forgiveness.universe, forgiveness[label].mf, label=label)

plt.title('Membership Functions for Input "Forgiveness"')
plt.xlabel('forgiveness level')
plt.ylabel('Membership Degree')
plt.legend()
plt.show()
```

```python
plt.figure()
for label in stochastic.terms:
    plt.plot(stochastic.universe, stochastic[label].mf, label=label)

plt.title('Membership Functions for Input "Stochastic Behavior"')
plt.xlabel('Stochastic level')
plt.ylabel('Membership Degree')
plt.legend()
plt.show()
```

```python
plt.figure()
for label in resulting_strategy.terms:
    plt.plot(resulting_strategy.universe, resulting_strategy[label].mf, label=label)

plt.title('Membership Functions for Output "Resulting strategy"')
plt.xlabel('Resulting strategy level')
plt.ylabel('Membership Degree')
plt.legend()
plt.show()
```
