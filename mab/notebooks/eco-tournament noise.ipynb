```python
import axelrod as axl
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
p = [player() for player in axl.axelrod_first_strategies]

p.append(axl.Fuzzy())

t = axl.Tournament(players=p, repetitions=1)

results = t.play()
```

```python
eco = axl.Ecosystem(results)
eco.reproduce(100)

print("First")
plot = axl.Plot(results)
p = plot.stackplot(eco)
p.show()
```
