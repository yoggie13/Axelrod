```python
import axelrod as axl
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
players = [player() for player in axl.demo_strategies]

for p in players:
    match = axl.Match((axl.Fuzzy(), p))

    match.play()

    print(f"(Fuzzy, {p.name}) : {str(match.final_score())}")
print("")
for p in players:
    match = axl.Match((axl.TitForTat(), p))

    match.play()

    print(f"(Tit For Tat, {p.name}) : {str(match.final_score())}") 
```
