```python
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
import axelrod as axl
```

```python
p = [player() for player in axl.axelrod_first_strategies]


p.append(axl.Fuzzy())


t = axl.Tournament(players=p, repetitions=1, noise=0.1)

results = t.play()

eco = axl.Ecosystem(results)
eco.reproduce(10000)

print("First - 10% noise")
plot = axl.Plot(results)
p = plot.stackplot(eco)
p.show()
```
