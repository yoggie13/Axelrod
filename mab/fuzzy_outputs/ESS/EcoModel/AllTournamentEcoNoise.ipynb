```python
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
import axelrod as axl
```

```python
p = [player() for player in axl.filtered_strategies({         "long_run_time": False, })]


t = axl.Tournament(players=p, repetitions=1, noise=0.1)

results = t.play()

eco = axl.Ecosystem(results)
eco.reproduce(20000)


df = pd.DataFrame({
    "Strategy": [str(player) for player in t.players],
    "Final Population": eco.population_sizes[-1]
})
df = df.sort_values(by="Final Population", ascending=False)

df.to_excel("axelrod_final_population_noise_1.xlsx", index=False)
```
