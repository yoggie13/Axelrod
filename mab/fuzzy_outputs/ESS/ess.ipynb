```python
import axelrod as axl
import pandas as pd
```

```python
players = [player() for player in axl.filtered_strategies({
        "long_run_time": False,
})]
players.remove(axl.Darwin())
players.remove(axl.Fuzzy())

from collections import defaultdict

fail_counts = defaultdict(int)
neutral_counts = defaultdict(int)

for i in range(0,30):
    fails = []
    neutral = []

    for p in players:
        m = axl.Match((axl.Fuzzy(), p))
        m.play()

        if(m.final_score()[1] > 600):
            fail_counts[str(p)] += 1

        
        elif(m.final_score()[1] == 600):
            alt_m = axl.Match((p, p))
            alt_m.play()

            if(m.final_score()[0] < min(alt_m.final_score()[0], alt_m.final_score()[1])):
                fail_counts[str(p)] += 1
            elif(m.final_score()[0] == min(alt_m.final_score()[0], alt_m.final_score()[1])):
                neutral_counts[str(p)] += 1
```

```python
df_fails = pd.DataFrame(list(fail_counts.items()), columns=['Strategy', 'Count'])
df_neutrals = pd.DataFrame(list(neutral_counts.items()), columns=['Strategy', 'Count'])

# Sort by count descending
df_fails = df_fails.sort_values('Count', ascending=False).reset_index(drop=True)
df_neutrals = df_neutrals.sort_values('Count', ascending=False).reset_index(drop=True)

with pd.ExcelWriter(f"ESS.xlsx") as writer:
    df_fails.to_excel(writer, sheet_name="Fails", startrow=0)
    df_neutrals.to_excel(writer, sheet_name="Neutrals", startrow=0)

```

```python
all_strategy_names = [str(p) for p in players]
passed_strategies = [s for s in all_strategy_names if s not in fail_counts and s not in neutral_counts]
print(f"\nStrategies that passed all 30 runs: {len(passed_strategies)}")
print(passed_strategies)
```
