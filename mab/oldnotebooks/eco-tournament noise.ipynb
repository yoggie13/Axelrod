```python
import axelrod as axl
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
p = [player() for player in axl.axelrod_first_strategies]

p.append(axl.Fuzzy())

t = axl.Tournament(players=p, repetitions=1, noise=0.1)

results = t.play()
```

```python
eco = axl.Ecosystem(results)
eco.reproduce(20000)

print("First")
plot = axl.Plot(results)
p = plot.stackplot(eco)
p.show()
```

```python
dfs = []


for i in range(0, 5):
    p = [axl.Fuzzy() for _ in range(5)] + [axl.Defector() for _ in range(20)] + [axl.TitForTat() for _ in range(15)] + [axl.Random() for _ in range(10)]

    t = axl.Tournament(players=p)

    results = t.play()

    eco = axl.Ecosystem(results)
    eco.reproduce(100)

    plot = axl.Plot(results)
    p = plot.stackplot(eco)
    print("All strategies")
    p.show()

    df = pd.DataFrame({
        "Strategy": [str(player) for player in t.players],
        "Final Population": eco.population_sizes[-1]
    })
    df = df.sort_values(by="Final Population", ascending=False)

    dfs.append(df)
```

```python
# 5. Save to Excel
j = 0
for df in dfs:
    df.to_excel("axelrod_final_population_1610.xlsx", sheet_name=str(j), index=False)
    j += 1

```
