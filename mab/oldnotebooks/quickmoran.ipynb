```python
import axelrod as axl
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
p = [axl.Fuzzy() for _ in range(5)] + [axl.Defector() for _ in range(20)] + [axl.TitForTat() for _ in range(15)] + [axl.Random() for _ in range(10)]

win = []
```

```python
for i in range (0, 15):
    mp = axl.MoranProcess(players=p, stop_on_fixation=True)

    populations = mp.play()
    
    ax = mp.populations_plot()
    win.append(mp.winning_strategy_name)
    plt.show()    
```

```python
unique_winners = set(win)

for uw in unique_winners:
    print(uw, win.count(uw)/len(win)*100)
```
