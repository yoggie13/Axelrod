```python
# run_queue.py
import papermill as pm
from pathlib import Path

notebooks = [
    "pso 2.ipynb",
    "pso mamdani.ipynb",
]

out_dir = Path("executed")
out_dir.mkdir(exist_ok=True)

for nb in notebooks:
    print(f"Running {nb}...")
    try:
        pm.execute_notebook(
            nb,
            out_dir / nb,        # executed copy, with outputs saved
            kernel_name="python3",
            # parameters={"alpha": 0.1},  # optional, needs a "parameters"-tagged cell
        )
    except Exception as e:
        print(f"Failed: {nb}: {e}")
        break                    # remove this to continue past failures
```
